```python
# Exercise 3 
num = int(input("Enter a number?\n"))
i = 2
while i < num:
    if (num % i == 0):
        print(f"{num} is NOT a prime number")
        break
    i += 1
else: 
    print(f"{num} is a prime number")
  
```

```python
# Exercise 4
num = int(input("Enter a number?\n"))
divisors = [1]
for i in range (2, num):
    if (num % i == 0):
        divisors.append(i)
if sum(divisors) == num:
    print(f"{num} is a perfect number")
else:
    print(f"{num} is NOT a perfect number")

```

```python
# Exercise 5
colors = ["blue", "green", "pink", "purple", "orange", "yellow"]
favco = input("What is your favourite color?\n")
i = 0
while i < len(colors):
    if favco == colors[i]:
        print(f"Your color is at index {i} in my list")
        break
    i += 1
else:
    print("Sorry, I could not find your color")
```

```python
# Exercise 8
def extract_even(l):
    even_list = []
    for i in range(len(l)):
        if l[i] % 2 == 0:
            even_list.append(l[i])
    return even_list

l = [1,4,5,-1,10]
extract_even(l)
```

```python
# Exercise 9
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)

num = input("Enter a number: \n")
print(factorial(int(num)))
```

```python
# Exercise 10
divisors = []
n = int(input("Enter a number: \n"))
for i in range(1,n+1):
    if n % i == 0:
        divisors.append(i)
print(f"The divisors of {n} are: ", end ="")
for i in range(len(divisors)):
    print(divisors[i], end =" ")
```

```python
# Exercise 11
import math
x1, y1 = map(int, input("Enter the coordinates x1, y1 of the first point: ").split())
x2, y2 = map(int, input("Enter the coordinates x2, y2 of the second point: ").split())

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print(f"The distance between 2 points is {distance}")
```

```python
# Exercise 12
m, n = map(int, input("Enter 2 integers: ").split())
for i in range(m):
    for j in range(n):
        if i == 0 or i == m - 1 or j == 0 or j == n - 1:
            print("*", end = "  ")
        else: 
            print(" ", end = "  ")
    print()
```
